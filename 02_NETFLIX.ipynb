```python
#LOAD THE DATASET
import pandas as pd
import numpy as np 
df = pd.read_csv("C:/Users/aditi/OneDrive/Desktop/Netflix/netflix_titles_dataset.csv")
print(df.shape)
df.head()
```

```python
#INSPECT THE DATASET
print(df.info())
print(df.describe(include="all"))
print(df.isnull().sum())
```

```python
#REMOVE LEADING/TRAILING SPACES

# Remove unnecessary spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
#CHECK FOR DUPLICATE RECORDS
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate show IDs:", df["show_id"].duplicated().sum())

df = df.drop_duplicates()
df = df.drop_duplicates(subset="show_id")
```

```python
#HANDLE MISSING date_added
df["date_added"] = df["date_added"].replace(
    ["", " ", "NULL", "null", "N/A"],
    np.nan
)
```

```python
#CONVERT date_added INTO A PROPER DATE
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

print(df["date_added"].isnull().sum())
print(df["date_added"].dtype)
```

```python
#CREATE added_year, added_month AND added_month_number
df["added_year"] = df["date_added"].dt.year

df["added_month"] = df["date_added"].dt.month_name()

df["added_month_number"] = df["date_added"].dt.month
```

```python
#CREATE THE rating COLUMN
df["rating"] = df["rating"].replace(
    ["", " ", "NULL", "null", "N/A"],
    np.nan
)

df["rating"] = df["rating"].fillna("Unknown")
```

```python
#STANDARDIZE duration_type
df["duration_type"] = df["duration_type"].replace({
    "Seasons": "Season"
})
```

```python
#MAKE SURE duration IS NUMERIC
df["duration"] = pd.to_numeric(
    df["duration"],
    errors="coerce"
)

print(df["duration"].isnull().sum())
```

```python
#CREATE SEPARATE movie AND TV-show duration COLUMNS
df["movie_duration"] = np.where(
    df["type"] == "Movie",
    df["duration"],
    np.nan
)

df["tv_show_seasons"] = np.where(
    df["type"] == "TV Show",
    df["duration"],
    np.nan
)
```

```python
#STANDARDIZE type
df["type"] = df["type"].str.strip().str.title()
```

```python
#CLEAN THE title COLUMN
df["title"] = df["title"].str.strip()

df["title"] = df["title"].replace(
    ["", " ", "NULL", "null"],
    np.nan
)

print(df["title"].isnull().sum())
```

```python
#VALIDATE release_year
df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
).astype("Int64")

print(df["release_year"].min())
print(df["release_year"].max())
```

```python
#CHECK FOR INCONSISTENT VALUES
print("Types:")
print(df["type"].unique())

print("\nRatings:")
print(df["rating"].unique())

print("\nDuration types:")
print(df["duration_type"].unique())
```

```python
#FINAL MISSING-VALUE CHECK
print(df.isnull().sum())
```

```python
#FINAL DATASET INFORMATION
print(df.info())
print(df.shape)
```

```python
#SAVE THE CLEANED DATASET
df.to_csv(
    "netflix_titles_dataset_cleaned.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")
```
